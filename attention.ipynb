```python
def softmax(x):
    return (np.exp(x).T / np.sum(np.exp(x), axis=1)).T

def scaled_dot_product_attention(q,k,v, mask=None):
    d_k = q.shape[-1]
    scaled = np.matmul(q, k.T)
    if mask is not None:
        scaled = scaled + mask
    attention = softmax(scaled)
    out = np.matmul(attention, v)
    return out, attention
```

```python
def scaled_dot_product(q,k,v, mask=None):
    d_k = q.size()[-1]
    scaled = torch.matmul(q, k.transpose(-1,-2)) / math.sqrt(d_k)
    if mask is not None:
        scaled += mask
    attention = F.softmax(scaled, dim=1)
    values = torch.matmul(attention, v)
    return values, attention
```

```python
tokens= ["Yo", "fui", "a", "Stanford", "y", "aprendi"]
```

```python
vocab = {
    "Yo" : 0,
    "fui" : 1,
    "a": 2,
    "Stanford" : 3,
    "y": 4,
    "aprendi": 5
}
```

```python
ids = [vocab[token] for token in tokens]
```

```python
ids
```

```python
vocab_size = 6
d_v = 4
```

```python
capa = nn.Embedding(vocab_size, d_v)
```

```python
tensor_ids = torch.tensor(ids, dtype=torch.long)
```

```python
x = capa(tensor_ids)
```

```python
x
```

```python
x.shape
```

```python
d_model = 4
d_h = 2

d_head = d_model // d_h

#Primera head de attention

w_q1 = nn.Linear(d_model,d_head, bias=False)
w_k1 = nn.Linear(d_model,d_head, bias=False)
w_v1 = nn.Linear(d_model,d_head, bias=False)
w_q2 = nn.Linear(d_model,d_head, bias=False)
w_k2 = nn.Linear(d_model,d_head, bias=False)
w_v2 = nn.Linear(d_model,d_head, bias=False)
```

```python
q1 = w_q1(x)
k1 = w_k1(x)
v1 = w_v1(x)
q2 = w_q2(x)
k2 = w_k2(x)
v2 = w_v2(x)
```

```python
scores1 = q1 @ k1.transpose(-2,-1)
scores1.shape
```

```python
scores2 = q2 @ k2.transpose(-2,-1)
scores2.shape
```

```python
scores1
```

```python
scores2
```

```python
scores_escalados = scores1 / (d_head ** 0.5)
scores_escalados2 = scores2 / (d_head ** 0.5)
```

```python
#Masking

mask = torch.triu(
    torch.ones_like(scores1, dtype=torch.bool),
    diagonal=1
)

scores_masked = scores_escalados.masked_fill(
    mask,float("-inf")
)

mask2 = torch.triu(
    torch.ones_like(scores2, dtype=torch.bool),
    diagonal=1
)

scores_masked2 = scores_escalados2.masked_fill(
    mask2,float("-inf")
)
```

```python
pesos1 = torch.softmax(scores_masked, dim=-1)
pesos2 = torch.softmax(scores_masked2, dim=-1)
```

```python
pesos1
```

```python
salida1= pesos1 @ v1
```

```python
salida2=pesos2 @ v2 
```

```python
salida = torch.cat(
    [salida1, salida2],
    dim=-1

)
```

```python
salida
```

```python
w_o = nn.Linear(4,4,bias=False)

#w_o es la matirz de pesos final
```

```python
output = w_o(salida)
```
